## Load dataset - MNLI

```python
mnli = load_dataset("nyu-mll/glue", "mnli")
```

## Analyse

```python
for split_name, dataset in mnli.items():
    print(f"{split_name}: {len(dataset):,} examples")
```

```python
mnli["train"].features
```

```python
example = mnli["train"][0]

example
```

```python
label_names = mnli["train"].features["label"].names
for i in range(0, 20, 4):
    example = mnli["train"][i]

    print(f"Example {i}")
    print("Premise:   ", example["premise"])
    print("Hypothesis:", example["hypothesis"])
    print("Label:     ", label_names[example["label"]])
    print("-" * 80)
```
