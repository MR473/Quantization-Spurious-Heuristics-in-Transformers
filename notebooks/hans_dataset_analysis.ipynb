# IMPORTS

```python
from datasets import load_dataset
import pandas as pd
```

# Load Dataset - HANS

```python
HANS_URL = (
    "https://raw.githubusercontent.com/tommccoy1/hans/"
    "master/heuristics_evaluation_set.txt"
)

hans = load_dataset(
    "csv",
    data_files={"validation": HANS_URL},
    delimiter="\t"
)

hans
```

# Dataset Structure

```python
hans["validation"].features
```

```python
hans["validation"][0]
```

# Dataset distribution

```python
print("Number of examples:", len(hans["validation"]))
print("Columns:", hans["validation"].column_names)
```

```python
df = hans["validation"].to_pandas()
print("Label distribution:\n")
print(df["gold_label"].value_counts())

print("Heuristic distribution:\n") 
print(df["heuristic"].value_counts())

print("\nNumber of subcases:", df["subcase"].nunique())
print("Number of templates:", df["template"].nunique())
```

# Heuristic and label distribution

```python
pd.crosstab(df["heuristic"], df["gold_label"])
```

```python
pd.crosstab(df["subcase"], df["gold_label"])
```
